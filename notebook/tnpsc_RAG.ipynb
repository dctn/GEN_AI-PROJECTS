```python
import os

import datasets
import pandas as pd
```

```python
dataset = datasets.load_dataset('snegha24/Tamil_tnpscExam')
```

```python
dataset.save_to_disk('../files/tnpsc')
```

```python
df = pd.DataFrame(dataset['train'])
df
```

```python
df.info()
```

```python
df['answer'] = df['answer'].astype(str)
```

```python
def extract_ans_from_list(series):
    try:
        ans_index = int(series['answer'])
        series['answer'] = series['options'][ans_index - 1]
    except:
        pass
    return series
```

```python
df = df.apply(extract_ans_from_list, axis=1)
df
```

# RAG

```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
docs = []

for i, row in df.iterrows():
    docs.append(
        Document(
            page_content=row['question'],
            metadata={'options': ", ".join(row['options']),
                      'answer':row['answer'],
                      'original_question_num': row['original_question_num'],
                      "link": row['source'],
                      'question_asked_paper':row['file_name']
                      },
        )
    )
```

## Embedding

```python
embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
embedding
```

## VECTOR DATABASE

```python
vector_db = Chroma().from_documents(docs,embedding=embedding,collection_name='TNPSC',persist_directory="./chroma_exam_db")
```

```python
responses = vector_db.similarity_search_with_score('எபிதீலியல்‌ செல்லில்‌ புற்றுநோய்‌ உருவாவதற்கு ________ என்று பெயர்‌.',k=10)
```

```python
for index in range(len(responses)):
    print(responses[index])
```
